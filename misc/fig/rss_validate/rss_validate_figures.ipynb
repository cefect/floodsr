# RSS Validation Figure Export

Load prepared rasters from `data/` and export report-style PNG figures for RSS Dudelange A and RSS Mersch A.

```python
# Load prepared case manifests and plotting helpers.
import json
from pathlib import Path

import matplotlib.pyplot as plt

from rss_validate import CASE_D, load_case_data, plot_comparison, plot_input_diagnostics

FIG_DIR = Path.cwd()
DATA_DIR = FIG_DIR / "data"
OUT_DIR = FIG_DIR / "output"
OUT_DIR.mkdir(exist_ok=True)
root_manifest = json.loads((DATA_DIR / "manifest.json").read_text())
case_names = list(CASE_D)
print(f"data_dir={DATA_DIR}")
print(f"cases={case_names}")
```

## Build Figures

```python
# Generate and save both report-style figures for each validation case.
metrics_df_d = {}
for case_name in case_names:
    case_d = load_case_data(DATA_DIR, case_name)
    print(f"{case_name}: costgrow_available={case_d['manifest']['params']['costgrow_available']}")
    print(f"{case_name}: prediction_shape={case_d['aligned']['resunet_prediction'].shape}")

    fig = plot_input_diagnostics(case_d)
    fig.savefig(OUT_DIR / f"{case_name}_input_diagnostics.png", dpi=300, bbox_inches="tight")
    plt.close(fig)

    fig, metrics_df = plot_comparison(case_d)
    fig.savefig(OUT_DIR / f"{case_name}_comparison.png", dpi=300, bbox_inches="tight")
    plt.close(fig)
    metrics_df_d[case_name] = metrics_df.round(6)
```

## Metrics

```python
# Show the metric annotations used in the saved comparison figures.
for case_name, metrics_df in metrics_df_d.items():
    print(case_name)
    display(metrics_df)
```
